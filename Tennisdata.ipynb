COMPETITIONS QUERIES

```python
#1.#List all competitions along with their category name

mycursor.execute(
    '''SELECT c.competition_name,cat.category_name
FROM competitions c JOIN categories cat
ON c.category_id = cat.category_id;
    
    '''             
)
results = mycursor.fetchall()
headers = ["Competition Name", "Category Name"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```

```python
#2.Count the number of competitions in each category

mycursor.execute(
    '''SELECT Categories.category_name, COUNT(Competitions.competition_id) AS competition_count
FROM Competitions
JOIN Categories ON Competitions.category_id = Categories.category_id
GROUP BY Categories.category_name;
    '''             
)
results = mycursor.fetchall()
headers = ["category_name", "num_of_competitions"]
table = tabulate(results, headers, tablefmt="grid")
print(table)






```

```python
#3.Find all competitions of type 'doubles':

mycursor.execute(
    '''SELECT competition_name,type FROM Competitions WHERE type = 'doubles';
'''             
)
results = mycursor.fetchall()
headers = ["competition_name", "Type"]
table = tabulate(results, headers, tablefmt="grid")
print(table)

```

```python
#4.Get competitions that belong to a specific category (e.g., ITF Men)

mycursor.execute(
    '''SELECT c.competition_name, cat.category_name
FROM competitions c
JOIN categories cat ON c.category_id = cat.category_id
WHERE cat.category_name = 'ITF Men'; '''             
)
results = mycursor.fetchall()
headers = ["competition_name", "category_name"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```

```python
#5.Identify parent competitions and their sub-competitions

mycursor.execute(
    '''SELECT a.competition_id, a.parent_id, b.category_name
FROM competitions a JOIN categories b ON a.category_id = b.category_id'''             
)

results = mycursor.fetchall()
headers = ["sub_competition","parent_competition","category_name"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```

```python
#6.Analyze the distribution of competition types by category

mycursor.execute(
    '''SELECT Categories.category_name, Competitions.type, COUNT(Competitions.competition_id) AS count
FROM Competitions
JOIN Categories ON Competitions.category_id = Categories.category_id
GROUP BY Categories.category_name, Competitions.type;

'''
)
results=mycursor.fetchall()
headers=["category_name","competitions_types","num_of_competitions"]
table=tabulate(results,headers,tablefmt="grid")
print(table)
```

```python
#7.List all competitions with no parent (top-level competitions)
mycursor.execute(
    '''SELECT * FROM competitions WHERE parent_id="NOT"'''             
)
results = mycursor.fetchall()
headers = ["competition_id", "competition_name","parent_id","type","gender","category_id"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```

COMPETITOR RANKINGS QUERIES


```python
from tabulate import tabulate
```

```python
#1.Get all competitors with their rank and points:

mycursor.execute(
'''SELECT c.competitor_id, c.name, r.ranks, r.points
FROM Competitors c
JOIN Competitor_Rankings r ON c.competitor_id = r.competitor_id;'''
)

results = mycursor.fetchall()
headers = ["competitor_id","competitor_name","ranks","points"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```

```python
#2.Find competitors ranked in the top 5:

mycursor.execute(
'''SELECT c.competitor_id, c.name, r.ranks, r.points
FROM Competitors c
JOIN Competitor_Rankings r ON c.competitor_id = r.competitor_id
WHERE r.ranks <= 5
ORDER BY r.ranks;'''
)

results = mycursor.fetchall()
headers = ["competitor_id","competitor_name","ranks","points"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```

```python
#3.List competitors with no rank movement (stable rank):

mycursor.execute(
'''SELECT c.competitor_id, c.name, r.ranks, r.movement
FROM Competitors c
JOIN Competitor_Rankings r ON c.competitor_id = r.competitor_id
WHERE r.movement = 0;'''
)

results = mycursor.fetchall()
headers = ["competitor_id","competitor_name","ranks","movements"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```

```python
#4.Get the total points of competitors from a specific country (e.g., Croatia):

mycursor.execute(
'''SELECT c.country, SUM(r.points) as total_points
FROM Competitors c
JOIN Competitor_Rankings r ON c.competitor_id = r.competitor_id
WHERE c.country = 'Croatia';'''
)

results = mycursor.fetchall()
headers = ["country","total_points"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```

```python
#5.Count the number of competitors per country:

mycursor.execute(
'''SELECT c.country, COUNT(c.competitor_id) as num_competitors
FROM Competitors c
GROUP BY c.country;'''
)

results = mycursor.fetchall()
headers = ["country","num_of_competitors"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```

```python
#6.Find competitors with the highest points in the current week:

mycursor.execute(
'''SELECT c.competitor_id, c.name, r.ranks, r.points
FROM Competitors c
JOIN Competitor_Rankings r ON c.competitor_id = r.competitor_id
ORDER BY r.points DESC
LIMIT 1;
'''
)

results = mycursor.fetchall()
headers = ["competitior_id","competitor_name","rank","highest_point"]
table = tabulate(results, headers, tablefmt="grid")
print(table)
```
